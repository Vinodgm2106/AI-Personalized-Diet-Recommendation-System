```python
# =====================================================
# FEATURE ENGINEERING
# =====================================================

import pandas as pd

from sklearn.preprocessing import LabelEncoder

# =====================================================
# LOAD FINAL DATASET
# =====================================================

df = pd.read_csv(
    r"/home/cnssec/Downloads/diet/Cleaned_data/final_ai_diet_dataset.csv"
)

print("Original Shape:")
print(df.shape)
```

```python
# =====================================================
# CREATE BMI STATUS
# =====================================================

def bmi_goal(category):

    if category in ['Extremely Weak', 'Weak']:

        return 'Weight Gain'

    elif category == 'Normal':

        return 'Maintenance'

    else:

        return 'Weight Loss'

df['diet_goal'] = df['bmi_category'].apply(
    bmi_goal
)
```

```python

# =====================================================
# CREATE CALORIE NEED CATEGORY
# =====================================================

def calorie_category(calories):

    if calories < 1800:

        return 'Low Calorie'

    elif calories < 2500:

        return 'Medium Calorie'

    else:

        return 'High Calorie'

df['calorie_category'] = df['calories'].apply(
    calorie_category
)

```

```python
# =====================================================
# CREATE PROTEIN LEVEL
# =====================================================

def protein_category(protein):

    if protein < 10:

        return 'Low Protein'

    elif protein < 25:

        return 'Medium Protein'

    else:

        return 'High Protein'

df['protein_category'] = df['protein'].apply(
    protein_category
)
```

```python
# =====================================================
# CREATE BMI RANGE
# =====================================================

def bmi_range(bmi):

    if bmi < 18.5:

        return 'Underweight'

    elif bmi < 25:

        return 'Normal'

    elif bmi < 30:

        return 'Overweight'

    else:

        return 'Obese'

df['bmi_range'] = df['bmi'].apply(
    bmi_range
)
```

```python
# =====================================================
# DAILY WATER REQUIREMENT
# =====================================================

df['water_intake_liters'] = (
    df['weight'] * 0.033
).round(2)

# =====================================================
# IDEAL PROTEIN REQUIREMENT
# =====================================================

df['daily_protein_need'] = (
    df['weight'] * 1.6
).round(2)
```

```python
# =====================================================
# LABEL ENCODING
# =====================================================

encoder = LabelEncoder()

categorical_columns = [
    'bmi_category',
    'diet_goal',
    'calorie_category',
    'protein_category',
    'bmi_range',
    'exercise_level'
]

for column in categorical_columns:

    df[column] = encoder.fit_transform(
        df[column]
    )
```

```python
# =====================================================
# FINAL DATASET INFO
# =====================================================

print("\nNEW FEATURES ADDED")

print(df.head())

print("\nFINAL SHAPE")
print(df.shape)

print("\nCOLUMNS")
print(df.columns)
```

```python
# =====================================================
# SAVE FEATURE ENGINEERED DATASET
# =====================================================

output_path = (
    r"/home/cnssec/Downloads/diet/Cleaned_data/feature_engineered_dataset.csv"
)

df.to_csv(
    output_path,
    index=False
)

print("\nFEATURE ENGINEERING COMPLETED")
```
