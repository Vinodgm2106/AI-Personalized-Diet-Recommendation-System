```python
# =====================================================
# FINAL DATASET MERGING
# =====================================================

import pandas as pd

# =====================================================
# LOAD DATASETS
# =====================================================

bmi_df = pd.read_csv(
    r"/home/cnssec/Downloads/diet/Cleaned_data/cleaned_bmi_dataset.csv"
)

calorie_df = pd.read_csv(
    r"/home/cnssec/Downloads/diet/Cleaned_data/cleaned_calorie_dataset.csv"
)

nutrition_df = pd.read_csv(
    r"/home/cnssec/Downloads/diet/Cleaned_data/cleaned_nutrition_dataset.csv"
)

print("complete")
```

```python
# =====================================================
# DISPLAY SHAPES
# =====================================================

print("BMI Dataset:", bmi_df.shape)

print("Calorie Dataset:", calorie_df.shape)

print("Nutrition Dataset:", nutrition_df.shape)

```

```python
# =====================================================
# COMBINE BMI + CALORIE DATASET
# =====================================================

combined_df = pd.concat(
    [bmi_df, calorie_df],
    axis=1
)
```

```python
# =====================================================
# REMOVE DUPLICATE COLUMNS
# =====================================================

combined_df = combined_df.loc[
    :,
    ~combined_df.columns.duplicated()
]
```

```python
# =====================================================
# RANDOMLY ATTACH FOOD DATA
# =====================================================

nutrition_sample = nutrition_df.sample(
    n=len(combined_df),
    replace=True,
    random_state=42
).reset_index(drop=True)
```

```python
# =====================================================
# FINAL MERGE
# =====================================================

final_df = pd.concat(
    [combined_df, nutrition_sample],
    axis=1
)
```

```python
# =====================================================
# REMOVE DUPLICATE COLUMNS AGAIN
# =====================================================

final_df = final_df.loc[
    :,
    ~final_df.columns.duplicated()
]


```

```python

# =====================================================
# HANDLE NULL VALUES
# =====================================================

for column in final_df.columns:

    if final_df[column].dtype == 'object':

        final_df[column] = final_df[column].fillna('Unknown')

    else:

        final_df[column] = final_df[column].fillna(0)
```

```python
# =====================================================
# FINAL DATASET INFO
# =====================================================

print("\nFINAL DATASET")
print(final_df.head())

print("\nFINAL SHAPE")
print(final_df.shape)

print("\nCOLUMNS")
print(final_df.columns)
```

```python
# =====================================================
# SAVE FINAL DATASET
# =====================================================

output_path = (
    r"/home/cnssec/Downloads/diet/Cleaned_data/final_ai_diet_dataset.csv"
)

final_df.to_csv(
    output_path,
    index=False
)

print("\nFINAL DATASET CREATED SUCCESSFULLY")
```
