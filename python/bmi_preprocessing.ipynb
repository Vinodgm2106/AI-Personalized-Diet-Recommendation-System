```python
# =====================================================
# BMI DATASET PREPROCESSING
# =====================================================

import pandas as pd

# =====================================================
# LOAD DATASET
# =====================================================

bmi_path = r"/home/cnssec/Downloads/diet/datasets/BMI_CAL/bmi.csv"

bmi_df = pd.read_csv(bmi_path)

print("Original Dataset Shape:")
print(bmi_df.shape)
```

```python
# =====================================================
# KEEP IMPORTANT COLUMNS
# =====================================================

bmi_df = bmi_df[
    ['Gender', 'Height', 'Weight', 'Index']
]
```

```python
# =====================================================
# REMOVE NULL VALUES
# =====================================================

bmi_df.dropna(inplace=True)

# =====================================================
# REMOVE DUPLICATES
# =====================================================

bmi_df.drop_duplicates(inplace=True)
```

```python
# =====================================================
# RENAME COLUMNS
# =====================================================

bmi_df.rename(columns={
    'Gender': 'gender',
    'Height': 'height',
    'Weight': 'weight',
    'Index': 'bmi_category'
}, inplace=True)
```

```python
# =====================================================
# ENCODE GENDER
# =====================================================

bmi_df['gender'] = bmi_df['gender'].map({
    'Male': 1,
    'Female': 0
})
```

```python
# =====================================================
# BMI CATEGORY MAPPING
# =====================================================

bmi_mapping = {
    0: 'Extremely Weak',
    1: 'Weak',
    2: 'Normal',
    3: 'Overweight',
    4: 'Obesity',
    5: 'Extreme Obesity'
}

bmi_df['bmi_category'] = (
    bmi_df['bmi_category']
    .map(bmi_mapping)
)
```

```python
# =====================================================
# CALCULATE BMI
# =====================================================

bmi_df['height_m'] = bmi_df['height'] / 100

bmi_df['bmi'] = (
    bmi_df['weight'] /
    (bmi_df['height_m'] ** 2)
)
```

```python
# =====================================================
# DROP TEMP COLUMN
# =====================================================

bmi_df.drop(columns=['height_m'], inplace=True)
```

```python
# =====================================================
# FINAL OUTPUT
# =====================================================

print("\nProcessed Dataset:")
print(bmi_df.head())

print("\nFinal Shape:")
print(bmi_df.shape)
```

```python
# =====================================================
# SAVE DATASET
# =====================================================

output_path = (
    r"/home/cnssec/Downloads/diet/Cleaned_data/cleaned_bmi_dataset.csv"
)

bmi_df.to_csv(
    output_path,
    index=False
)

print("\nBMI preprocessing completed")
```
