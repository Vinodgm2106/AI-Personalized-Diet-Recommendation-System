```python
# =====================================================
# USER NUTRITION DATASET PREPROCESSING
# =====================================================

import pandas as pd

# =====================================================
# LOAD DATASET
# =====================================================

calorie_path = (
    r"/home/cnssec/Downloads/diet/datasets/daily_nutrition/user_nutritional_data.csv"
)

calorie_df = pd.read_csv(calorie_path)

print("Original Dataset Shape:")
print(calorie_df.shape)
```

```python

# =====================================================
# KEEP IMPORTANT COLUMNS
# =====================================================

calorie_df = calorie_df[
    [
        'Age',
        'Daily meals frequency',
        'Physical exercise',
        'BMR',
        'Carbs',
        'Proteins',
        'Fats',
        'Calories'
    ]
]
```

```python
# =====================================================
# REMOVE NULL VALUES
# =====================================================

calorie_df.dropna(inplace=True)

# =====================================================
# REMOVE DUPLICATES
# =====================================================

calorie_df.drop_duplicates(inplace=True)
```

```python
# =====================================================
# RENAME COLUMNS
# =====================================================

calorie_df.rename(columns={
    'Daily meals frequency': 'meals_per_day',
    'Physical exercise': 'exercise_level',
    'Carbs': 'carbs',
    'Proteins': 'proteins',
    'Fats': 'fats',
    'Calories': 'calories',
    'BMR': 'bmr',
    'Age': 'age'
}, inplace=True)
```

```python
# =====================================================
# FINAL OUTPUT
# =====================================================

print("\nProcessed Dataset:")
print(calorie_df.head())

print("\nFinal Shape:")
print(calorie_df.shape)
```

```python
# =====================================================
# SAVE CLEAN DATASET
# =====================================================

output_path = (
    r"/home/cnssec/Downloads/diet/Cleaned_data/cleaned_calorie_dataset.csv"
)

calorie_df.to_csv(
    output_path,
    index=False
)

print("\nCalorie preprocessing completed")
```
